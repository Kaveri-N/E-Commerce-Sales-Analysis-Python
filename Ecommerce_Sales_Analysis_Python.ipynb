# E-Commerce Sales Analysis using Python

## Project Objective

The objective of this project is to analyze e-commerce sales data using Python and identify important business trends and patterns.

The analysis focuses on revenue performance, customer segments, sales channels, product categories, product performance, return rates, and the impact of discounts.

The insights generated from the analysis can help businesses improve sales performance, customer targeting, product strategy, and overall decision-making.

## Dataset Overview

The dataset contains 8,000 e-commerce order records.

### Key Columns

- **order_date** – Date of the order
- **order_id** – Unique order identifier
- **customer_id** – Customer identifier
- **customer** – Customer name
- **segment** – Customer segment
- **channel** – Sales/marketing channel
- **product** – Product name
- **category** – Product category
- **quantity** – Quantity purchased
- **unit_price** – Price per unit
- **discount_pct** – Discount percentage
- **line_total** – Total value of the order line
- **returned** – Return indicator

The dataset was analyzed using Python, Pandas, NumPy, Matplotlib, and Seaborn.

## Data Cleaning & Quality Check

The dataset was checked and prepared before performing the analysis.

### Data Cleaning Steps

- Converted `order_date` from text format to datetime format.
- Checked for missing values — no missing values were found.
- Checked for duplicate records — no duplicate records were found.
- Verified that quantity and unit price values were positive.
- Verified that discount percentages were within a valid range.
- Validated `line_total` against quantity, unit price, and discount. All records were within acceptable rounding differences.
- Checked the `returned` column to calculate the overall return rate.

### Data Quality Result

The dataset was clean and suitable for further exploratory data analysis.

## Exploratory Data Analysis (EDA)

Exploratory Data Analysis was performed to understand the sales performance and identify important business patterns.

### Analysis Performed

- Calculated total revenue and total orders.
- Calculated Average Order Value (AOV).
- Analyzed monthly revenue trends.
- Compared revenue across product categories.
- Analyzed revenue by sales channels.
- Compared revenue across customer segments.
- Identified the Top 10 products by revenue.
- Analyzed return rates across product categories.
- Studied the relationship between discounts and revenue.

These analyses helped identify key revenue drivers, customer patterns, product performance, and potential areas for business improvement.

## Key Business Insights

The analysis revealed the following key business insights:

1. The business generated a total revenue of **₹4,87,706.66** from **4,400 orders**.

2. The **Average Order Value (AOV)** was **₹110.84**, indicating the average revenue generated per order.

3. **Electronics** was the highest revenue-generating product category.

4. **Email** was the highest revenue-generating sales channel.

5. The **Loyal** customer segment contributed the highest revenue.

6. **USB-C Cable** was the top-performing product by revenue.

7. The overall product return rate was **5.48%**.

8. **Electronics** had the highest category-level return rate at **5.83%**.

## Business Recommendations

Based on the analysis, the following recommendations can help improve business performance:

1. **Focus on Electronics:**  
   Electronics is the highest revenue-generating category. The business should maintain sufficient inventory and continue targeted marketing for high-performing electronic products.

2. **Strengthen Email Marketing:**  
   Email generates the highest revenue among sales channels. Personalized email campaigns and promotional offers can be used to increase customer engagement and sales.

3. **Retain Loyal Customers:**  
   The Loyal customer segment contributes the highest revenue. Loyalty programs, personalized offers, and repeat-purchase campaigns can help retain these customers.

4. **Promote Top-performing Products:**  
   High-revenue products such as the USB-C Cable should be kept readily available and promoted through effective sales campaigns.

5. **Reduce Product Returns:**  
   Electronics has the highest return rate at 5.83%. Product descriptions, quality checks, packaging, and customer expectations should be reviewed to reduce returns.

6. **Optimize Discount Strategy:**  
   Discount levels should be evaluated based on their impact on revenue to ensure that promotions increase sales without unnecessarily reducing profitability.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\Kaveri Nikam\Downloads\ecommerce_orders.csv")
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
```

```python
monthly_revenue = (
    df.groupby(df["order_date"].dt.to_period("M"))["line_total"]
      .sum()
      .reset_index()
)

monthly_revenue["order_date"] = monthly_revenue["order_date"].astype(str)

monthly_revenue
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    monthly_revenue["order_date"],
    monthly_revenue["line_total"],
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
category_revenue = (
    df.groupby("category")["line_total"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)
category_revenue
```

```python
plt.figure(figsize=(9,5))

plt.bar(
    category_revenue["category"],
    category_revenue["line_total"]
)

plt.title("Revenue by Product Category")
plt.xlabel("Category")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
channel_revenue = (
    df.groupby("channel")["line_total"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)
channel_revenue
```

```python
plt.figure(figsize=(9,5))

plt.bar(
    channel_revenue["channel"],
    channel_revenue["line_total"]
)

plt.title("Revenue by Sales Channel")
plt.xlabel("Sales Channel")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
segment_revenue = (
    df.groupby("segment")["line_total"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)

segment_revenue
```

```python
plt.figure(figsize=(9,5))

plt.bar(
    segment_revenue["segment"],
    segment_revenue["line_total"]
)

plt.title("Revenue by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Revenue (₹)")

plt.tight_layout()
plt.show()
```

```python
top_products = (
    df.groupby("product")["line_total"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
      .reset_index()
)

top_products
```

```python
plt.figure(figsize=(10,6))

plt.barh(
    top_products["product"],
    top_products["line_total"]
)

plt.title("Top 10 Products by Revenue")
plt.xlabel("Revenue (₹)")
plt.ylabel("Product")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
category_return_rate = (
    df.groupby("category")["returned"]
      .mean()
      .mul(100)
      .sort_values(ascending=False)
      .reset_index()
)

category_return_rate
```

```python
plt.figure(figsize=(9,5))

plt.bar(
    category_return_rate["category"],
    category_return_rate["returned"]
)

plt.title("Return Rate by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Return Rate (%)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
discount_revenue = (
    df.groupby("discount_pct")["line_total"]
      .sum()
      .reset_index()
      .sort_values("discount_pct")
)

discount_revenue
```

```python
plt.figure(figsize=(9,5))

plt.bar(
    discount_revenue["discount_pct"].astype(str),
    discount_revenue["line_total"]
)

plt.title("Revenue by Discount Percentage")
plt.xlabel("Discount (%)")
plt.ylabel("Revenue (₹)")

plt.tight_layout()
plt.show()
```
